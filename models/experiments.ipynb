```python
from StockModels import StockModels
import simulation
import numpy as np
from performance_metrics import calculate_model_performance, print_performance_report, compare_models, calculate_growth_percentage, calculate_test_performance, print_test_performance, compare_test_models
```

```python
models = StockModels()
```

```python
df_FV = models.load_data(use_roberta=False)
df_RB = models.load_data(use_roberta=True)
```

```python
features_FV = ["finvader_tot", "pos_art_count", "total_articles", "Open_Diff", "y", "Open"]
features_RB = ["frob_comp", "pos_art_count", "total_articles", "Open_Diff", "y", "Open"]
```

# Validation

## 1. Logistic Regression

```python
# Run Logistic Regression for FinVader
cv_trades_finvader, cv_opens_finvader, dumb_trades_finvader, accuracy_scores_finvader, rmse_scores_finvader = models.logistic_regression(df_FV, features_FV, val=True)
cv_trades_roberta, cv_opens_roberta, dumb_trades_roberta, accuracy_scores_roberta, rmse_scores_roberta = models.logistic_regression(df_RB, features_RB, val=True)
```

```python
lr_scores_finvader = {}
dumb_scores_finvader = {}
lr_scores_roberta = {}
dumb_scores_roberta = {}

for i in range(4):
    lr_scores_finvader[i] = simulation.get_performance(cv_trades_finvader[i], cv_opens_finvader[i])
    dumb_scores_finvader[i]= simulation.get_performance(dumb_trades_finvader[i], cv_opens_finvader[i])
    lr_scores_roberta[i] = simulation.get_performance(cv_trades_roberta[i], cv_opens_roberta[i])
    dumb_scores_roberta[i] = simulation.get_performance(dumb_trades_roberta[i], cv_opens_roberta[i])
```

### FinVader Validation Performance per CV fold and the whole validation set

```python
lr_performance_finvader = calculate_model_performance(
    lr_scores_finvader, 
    cv_opens_finvader, 
    accuracy_scores_finvader, 
    rmse_scores_finvader, 
    "FinVader LogReg"
)
```

```python
print_performance_report(lr_performance_finvader, "FinVader LogReg Performance")
```

### Fine-Tuned RoBERTa Validation Performance per CV fold and the whole validation set

```python
lr_performance_roberta = calculate_model_performance(
    lr_scores_roberta, 
    cv_opens_roberta, 
    accuracy_scores_roberta, 
    rmse_scores_roberta, 
    "RoBERTa LogReg"
)
```

```python
print_performance_report(lr_performance_roberta, "RoBERTa LogReg Performance")
```
